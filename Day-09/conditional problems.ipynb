```python
n = int(input())
if n>0:
    if n%2==0:
        print(f" {n} is positive even num")
    else:
        print(f" {n} is positive odd num")
elif n<0:
    if n%2==0:
        print(f" {n} is negative even num")
    else:
        print(f" {n} is negative odd num")
else:
    print(0)
    
```
