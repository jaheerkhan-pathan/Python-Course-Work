```python
n=int(input())
for r in range(n):
    for c in range(n):
        if r==0 or c==n-1 or c==0 or r==n-1:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
n=int(input())
for r in range(n):
    #spaces
    for c in range(r):
        print(' ',end=' ')
    #stars
    for c in range(2*n-(2*r+1)):
        print('*',end=' ')
    print()
        
```

```python
# pyramid num
num = 1
n = 5
for r in range(n):
    for c in range(n-r):
        print(" ", end = " ")
    for c in range(2*r+1):
        print(num, end = " ")
    print()
    
```

```python
#rhombus
num = 1
n = 2
for r in range(n):
    for c in range(n-r):
        print(" ", end = " ")
    for c in range(2*r+1):
        print(num, end = " ")
    print()
n = 3
for r in range(n):
    #spaces
    for c in range(r):
        print(" ", end = " ")
    for c in range(2*n-(2*r+1)):
        print(num, end = " ")
    print()
```
