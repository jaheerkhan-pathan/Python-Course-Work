```python
#build-in functions.
#changing global value.
x=20
def f2():
    global x #(using global keyword for changing global value)
    x=x+10
f2()
print(x)
```

```python
#changing local value.
def outer():
    x=10
    def inner():
        nonlocal x  #(using nonlocal keyword for chaning local value)
        x=x+100
        print(x)
    inner()
outer()
```

```python
#pass by value.
def f (x):
    print(id(x))
    print(x)
a= 20
print(id(a))
f(a)
```

```python
#pass by object refference.
def f (x):
    x=[10,20,30]
    print(id(x))
    print(x)
a=[40,50]
print(id(a))
f(a)
```

```python
## Pass by Object reference using immutable data type
def f(x):
    x=20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
# Pass by Object reference using immutable data type - modification or reassign
def f(x):
    x.append(10)
    print(id(x))
    x=[10,20,30,40]
    print(id(x))
    print(x)
a=[1,2,3,4]
print(id(a))
f(a)
```

```python
#Recursive Function
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
a=int(input())
print(factorial(a))
```

```python
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

```python
#sum of natural numbers
def sum_of_natural_numbers(n):
    if n==2:
        return 2
    if n==1:
        return 1
    return n+sum_of_natural_numbers(n-2)
print(sum_of_natural_numbers(9))
```
