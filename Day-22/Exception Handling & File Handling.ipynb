```python
Exception handling: 
keywords: try, except,else, finally
syntax:
try :
    block of code where we have the error in our code
except:
    print(..............)
else:
    print(.............)
finally:
    print(...........)
```

```python
a = int(input())
b = int(input())
try:
    print(a/b)
except:
    print("Something went wrong")
else:
    print("there is no error")
finally:
    print("execution done")
```

```python
age = 14
try:
    if age<18:
        raise AgeError("Age is less")
except:
    print("code is not working")
```

```python
class AgeError(Exception):
    pass
try:
    if age<18:
        raise AgeError("Age is less")
except AgeError as a:
    print(a)
```
