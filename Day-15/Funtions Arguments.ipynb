```python
# 1. positional arguments
def addition1(a,b):
    print(a,b)
    return a+b
print(addition1(10,20))
```

```python
# 2. keyword arguments
def addition1(a,b):
    print(a,b)
    return a+b
c = 100
d = 55
print(addition1(b=c,a=d))
```

```python
# 3. default arguments
def addition1(a,b,c=0,d=0):
    return a+b+c+d
print(addition1(10,20))
print(addition1(10,20,30))
print(addition1(10,20,30,40))
```

```python
# 4. variable len arguments
def addition1(*var):
    print(type(var))
    #return sum(var)
    total = 0
    for num in var:
         total += num
    return total
#print(addition1(10,20,30,40))
print(addition1(1,2,3,4,5,6,7))
```

```python
# 5. variable len keyword arguments
def addition1(**var):
    print(type(var))
    #return sum(var)
    total = 0
    for num in var.values():
         total += num
    return total
#print(addition1(10,20,30,40))
print(addition1(a=1,b=2,c=3))
```
