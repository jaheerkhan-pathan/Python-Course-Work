```python
def generator():
    yield 1
    yield 2
    yield 3
    yield 4
r = generator()
print(next(r))
print(next(r))
print(next(r))
print(next(r))
for i in r:
    print(i)
```

```python
def generator(x):
    x = x+1
    yield x
    x+=10
    yield x
    x+=20
    yield x
    x+=30
    yield x
r = generator(10)
print(type(r))
for i in r:
    print(i)
```
